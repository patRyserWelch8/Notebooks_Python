```python
I = np.array([[1,0],[0,1]])
I
```

$inverse = \frac{1}{ad-bc}\begin{bmatrix}
d & -b\\ 
 -c & a
\end{bmatrix}$

```python
A
```

Step 0 : calculate the scalar value

```python
ad = (A[0,0] * A[1,1])
ad
```

```python
bc = (A[0,1] * A[1,0])
bc 
```

```python
ad-bc
```

```python
scalar = 1/(ad-bc)
scalar
```

```python
row_0 = [A[1,1],-A[0,1]]
row_1 = [-A[1,0],A[0,0]]
```

```python
prep_A = np.array([row_0, row_1])
prep_A
```

```python
inverse = scalar * prep_A
inverse
```

```python
np.linalg.inv(A)
```

Assume  the very low values are 0s. 

```python
np.dot(A, inverse)
```

## A more complex example

$ x_1 + 2.5 x_2 + 10 x_3 + 5 x_4 = 150$

$ 0.5x_1 + 20 x_2 + 1.2 x_3 + 0.45 x_4 = 200$

$ 23 x_1 + 10 x_2 +  34 x_4 = 1000$

$  9 x_2 + 5.2 x_3 + 4 x_3 + 7 x_4 = 200$

It becomes

$AX = B$




```python
A = np.array([[1,2.5,10,5],
              [0.5, 20, 1.2, 0.45],
              [23,10,0,34],
              [0, 9, 5.2,7]])
A
```

```python
B = [150, 200, 1000,200]
```

```python
inverse = np.linalg.inv(A)
inverse
```

$X = A^{-1}B$

```python
X = np.dot(inverse, B)
X
```

Let's verify the values of x and y

```python
A[0,0] * X[0] + A[0,1] * X[1] + A[0,2] * X[2] + A[0,3] * X[3]
```

```python
A[1,0] * X[0] + A[1,1] * X[1] + A[1,2] * X[2] + A[1,3] * X[3]
```

```python
A[2,0] * X[0] + A[2,1] * X[1] + A[2,2] * X[2] + A[2,3] * X[3]
```

```python
A[3,0] * X[0] + A[3,1] * X[1] + A[3,2] * X[2] + A[3,3] * X[3]
```
